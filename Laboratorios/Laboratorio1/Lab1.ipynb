<h1>Conjunto de datos sobre el rendimiento estudiantil y hábitos de estudio</h1>
<p>Conjunto de datos simulado que analiza cómo afectan los hábitos de estudio, el estilo de vida y la demografía</p>

```python
# Esta celda importa librerías para manejo de datos, álgebra y gráficos.
# Sirve para preparar el entorno antes del entrenamiento del modelo.
import os
import numpy as np
import pandas as pd
from matplotlib import pyplot
from mpl_toolkits.mplot3d import Axes3D  # Necesario para algunas gráficas 3D
%matplotlib inline

import matplotlib.pyplot as plt
```

```python

dataset = 'student_performance_dataset.csv'
print(f'Archivo a usar: {os.path.abspath(dataset)}')
```

```python


df = pd.read_csv(dataset, sep=',', header=None)

X = df.iloc[:, 2].values
y = df.iloc[:, 10].values                

m = y.size
n = X.shape

print(f'Número de muestras: {m}')   
print(f'Número de características: {n}')

print(f'Tamaño de X: {X.shape}')
print(f'Tamaño de y: {n}')

print(f'Tamaño del dataset: {df.shape}')
```

```python
print('Valores de Y:')
print(y)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color="#0d53c4", alpha=0.7, edgecolors='k', label='Datos originales')
plt.title('Gráfico de Dispersión de los Datos Originales', fontsize=16)
plt.xlabel('Horas de Estudio', fontsize=12)
plt.ylabel('Calificacion sobre 100', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

```python

n = len(X)

b_min = -20
b_max = 20
w_min = -20
w_max = 20
paso = 0.1  

mejor_acum = -1      
mejor_b = 0
mejor_w = 0
total_combinaciones = 0


b = b_min
while b <= b_max:

    w = w_min
    while w <= w_max:

        acum = 0  

        for j in range(n):
            y_pred = b + w * X[j]
            error = abs(y[j] - y_pred)
            umbral = 0.10 * abs(y[j])   

            if error < umbral:
                acum = acum + 1

        
        if acum > mejor_acum:
            mejor_acum = acum
            mejor_b = b
            mejor_w = w

        total_combinaciones = total_combinaciones + 1
        w = w + paso

    b = b + paso


print("--- Búsqueda completada ---")
print(f"Combinaciones evaluadas: {total_combinaciones}")
print(f"Mejor b (intercepto): {mejor_b:.4f}")
print(f"Mejor w (pendiente): {mejor_w:.4f}")
print(f"Puntos dentro del umbral: {mejor_acum} de {n}")


print("\n--- Inferencias con los mejores b y w encontrados ---")

nuevos_x = [1.5, 2, 3, 4, 5, 6, 2.5, 1.8, 3.5, 4.5]  

for i in range(len(nuevos_x)):
    xv = nuevos_x[i]
    prediccion = mejor_b + mejor_w * xv
    print(f"x = {xv}  ->  y predicho = {prediccion:.4f}")
```

```python
plt.figure(figsize=(15, 6))
plt.scatter(X, y, color='#3b82f6', alpha=0.7, edgecolors='k', label='Datos originales')

y_recta = mejor_b + mejor_w * X
plt.plot(X, y_recta, color='#ef4444', linewidth=3, label=f'Recta Ajustada (y = {mejor_b:.2f} + {mejor_w:.2f}x)')

plt.title('Grafico', fontsize=14)
plt.xlabel('Horas', fontsize=12)
plt.ylabel('Calificación', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
